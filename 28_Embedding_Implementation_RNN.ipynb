```python
import numpy as np

docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'modi ji ki jai',
        'inquilab zindabad']
```

```python
from keras.preprocessing.text import Tokenizer
tokenizer= Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
```

```python
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences , padding= 'post')
sequences
```

```python
from keras.models import Sequential 
from keras.layers import Dense , SimpleRNN , Flatten , Embedding
```

```python
model = Sequential()
model.add(Embedding(17,output_dim= 2, input_length= 5 )) # 17 is input words
model.summary()
```

```python
model.compile('adam', 'accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

##### Now using Embedding in Real DataSet

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
x_train = pad_sequences(x_train, padding='post', maxlen=150)
x_test = pad_sequences(x_test, padding= 'post', maxlen= 150)
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(Embedding(10000, output_dim=2, input_length=150))
model.add(SimpleRNN(32, activation='tanh',
          input_shape=(150, 1), return_sequences=False))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',
              optimizer='adam', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=5,  validation_data=(x_test, y_test))
```
