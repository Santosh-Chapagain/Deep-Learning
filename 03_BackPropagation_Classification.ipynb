```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame([[8 , 8 , 1], [7 , 9 , 1] , [6 , 10 , 1] , [5 , 7 , 0]] , columns=['cgpa',  'iq' , 'placed'])
```

```python
def initialize_parameters(layer_dim):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dim)

    for l in range(1 , L):
        parameters['w' + str(l)] = np.ones((layer_dim[l-1] , layer_dim[l]))*0.1
        parameters['b'+str(l)] = np.zeros((layer_dim[l] ,1))

    return parameters
```

```python
#Utility function
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def linear_forward(A_prev , w , b):
    z = np.dot(w.T , A_prev) + b
    A = sigmoid(z)
    return A
```

```python
def l_layer_forward(x , parameters):
    A = x
    L = len(parameters)//2
    
    for l in range(1 , L+1):
        A_prev = A
        wl = parameters['w'+str(l)]
        bl = parameters['b'+str(l)]
        A = linear_forward(A_prev , wl , bl)

    return A , A_prev

```

```python
def update_parameters(parameters , y , y_hat , A1 , x):
    parameters['w2'][0][0] = parameters['w2'][0][0] + 0.001 * (y - y_hat) * A1[0][0]
    parameters['w2'][1][0] = parameters['w2'][1][0] + 0.001 * (y - y_hat) * A1[1][0]
    parameters['b2'][0][0] = parameters['b2'][0][0] + 0.001 * (y - y_hat)

    parameters['w1'][0][0] = parameters['w1'][0][0] + (0.001 * (y-y_hat) * parameters['w2'][0][0] * A1[0][0] * (1 - A1[0][0]) * x[0][0])
    parameters['w1'][0][1] = parameters['w1'][0][1] + (0.001 * (y-y_hat) * parameters['w2'][0][0] * A1[0][0] * (1 - A1[0][0]) * x[0][0])
    parameters['b1'][0][0] = parameters['w1'][0][0] + (0.001 * (y-y_hat) * parameters['w2'][0][0] * A1[0][0] * (1 - A1[0][0]))

    parameters['w1'][1][0] = parameters['w1'][1][0] + (0.001 * (y-y_hat) * parameters['w2'][1][0] * A1[1][0] * (1 - A1[1][0]) * x[0][0])
    parameters['w1'][1][1] = parameters['w1'][1][1] + (0.001 * (y-y_hat) * parameters['w2'][1][0] * A1[1][0] * (1 - A1[1][0]) * x[1][0])
    parameters['b1'][1][0] = parameters['w1'][1][0] + (0.001 * (y-y_hat) * parameters['w2'][0][0] * A1[1][0] * (1 - A1[1][0]))
```

```python
# epoch implementation
parameters = initialize_parameters([2, 2, 1])
epochs = 50

for i in range(epochs):
    loss = []
    for j in range(df.shape[0]):
        x = df[['cgpa', 'iq']].values[j].reshape(2, 1)
        y = df['placed'].values[j]

        y_hat, A1 = l_layer_forward(x, parameters)
        y_hat = y_hat[0][0]
        calc_loss = -(y*np.log(y_hat)) - (1-y)* np.log(1-y_hat)
        loss.append(calc_loss)
        update_parameters(parameters, y, y_hat, A1, x)

    print('Epoch - ', i+1, 'Loss - ', np.array(loss).mean())

parameters
```
